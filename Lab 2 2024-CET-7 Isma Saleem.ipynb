```python
import numpy as np

units =np.array(["Punjab", "Sindh", "KPK", "Balochistan", "Islamabad", "AJK"])
pop_2017 =np.array([109.99, 47.85, 35.50, 12.34, 2.00, 4.05])
pop_2023 =np.array([120, 105, 50, 38, 15, 3]) 
total =np.sum(pop_2023)
print("T6A: Total popoulation(Millions):", total)
share =(pop_2023/total)*100
print("\nT6B: Percentage share(%):")
print(np.round(share, 2))
most_populous =np.argmax(pop_2023)
print("\nT6C: Most populous unit:", units[most_populous])
above_20 =units[pop_2023>20]
print("\nT6D: Units above 20 millions:", above_20)
growth =((pop_2023/pop_2017)**(1/6)-1)*100
print("\nT6E: Anual growth rate(%):")
print(np.round(growth, 2))
fastest =np.argmax(growth)
print("\nT6F: Fastest growing units:", units[fastest])
print("growth rate(%):", round(growth[fastest], 2))
print("\nAJK Population comparison:")
print("2017:", pop_2017[5], "Million")
print("2023:", pop_2023[5], "Million")
```

```python
import numpy as np

cities =np.array(["Lahore", "karachi", "Islamabad", "Peshawar", "Quetta"])
months =np.array(["jan", "feb", "mar", "apr", "may", "june", "july", "aug", "sep" ,"oct", "nov", "dec"])
temp =np.array([
    [13, 16, 22, 28, 33, 34, 32, 31, 30, 26, 20, 15],
    [19, 21, 25, 29, 31, 31, 30, 29, 29, 28, 24, 20],
    [10, 13, 18, 24, 29, 31, 29, 28, 27, 23, 17, 12],
    [11, 14, 20, 27, 33, 35, 34, 33, 30, 25, 18, 12],
    [4, 7, 12, 18, 23, 29, 30, 28, 24, 18, 11, 6]
])
print("T7A: Temperature array shape:", temp.shape)
yearly_mean =np.mean(temp, axis=1)
print("\nT7B: yearly mean temperature:")
print(np.round(yearly_mean, 2))
hottest =np.argmax(temp, axis=1)
print("\nT7C: hottest month:")
print(months[hottest])
monthly_avg =np.mean(temp, axis=0)
print("\nT7D: five city monthly average:")
print(np.round(monthly_avg, 2))
hot_months =months[np.any(temp>35, axis=0)]
print("\nT7E: months above 35:")
print(hot_months)
Lahore_f =(temp[0]*9/5) +32
print("\nT7F: Lahore temperature in fahrenheit:")
print(np.round(Lahore_f, 2))
rainfall =np.array([
    [180, 165, 86],
    [60, 70, 35],
    [350, 300, 250],
    [100, 90, 50],
    [40, 35, 20]
])
monsoon_total =np.sum(rainfall, axis=1)
print("\nT7G: Monsoon rainfall(mm):")
print(monsoon_total)
wettest =np.argmax(monsoon_total)
print("\nT7H: Wettest Monsoon city:", cities[wettest])
monsoon_share =(monsoon_total/np.sum(monsoon_total))*100
print("monsoon share(%):")
print(np.round(monsoon_share, 2))
    

    
```

```python
import numpy as np

months =np.array(["july", "august", "september", "october", "november", "december", "january", "febuary", "march", "april", "may", "june"])
fy2024 =np.array([
    28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6
])
fy2025 =np.array([11.1, 9.6, 6.9, 7.2, 4.9, 4.1])
avg_2024 =np.mean(fy2024)
avg_2025 =np.mean(fy2025)
print("T8A: fy2024 average inflation(%):", round(avg_2024, 2))
print("fy2025 average inflation(%):", round(avg_2025, 2))
drop =fy2024[:-1] - fy2024[1:]
print("\nT8B: monthly inflation drop:")
print(np.round(drop, 2))
biggest_drop =np.argmax(drop)
print("\nT8C: Biggest drop after:", months[biggest_drop])
category =np.where(fy2025<10, "low", "moderate")
print("\nT8D: fy2025 inflation categories:")
print(category)
print("T8E: Inflation interpretation:")
print("inflation decreased during fy2025")
print("lower inflation means prices are increasing more slowly")
```

```python
import numpy as np

marks =np.array([35, 42, 48, 28, 55, 32, 60, 38, 25, 70])
grace_marks =np.clip(marks +5, 0, 100)
print("\nT9A: Marks after grace:")
print(grace_marks)
passed =grace_marks>=40
pass_percentage =np.mean(passed)*100
print("\nT9B: pass percentage(%):", pass_percentage)
grace_only =(marks<40)&(grace_marks>=40)
print("\nT9C: Passed only because of grace:")
print(np.where(grace_only)[0]+1)
z_scores =(grace_marks - np.mean(grace_marks))/np.std(grace_marks)
print("\nT9D: z_scores:")
print(np.round(z_scores, 2))
```

```python
import numpy as np

image =np.zeros((8, 8), dtype=np.uint8)
image[1, :] =255
image[2:7, 6]=255
print("T10A: original image:")
print(image)
flipped =np.fliplr(image)
print("\nT10B: Flipped image:")
print(flipped)
inverter =255-image
print("\nT10C: inverted image:")
print(inverter)
cropped =image[1:7, 1:7]
print("\nT10D: Cropped image:")
print(cropped)
brightened =np.clip(image+100, 0, 255)
print("\nT10E: Brightened image:")
print(brightened)
print("\nT10F: Image shape:", image.shape)
print("image data type:", image.dtype)
print("white pixels:", np.sum(image==255))
```
