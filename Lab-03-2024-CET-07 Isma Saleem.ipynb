 ## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# Missing values
import pandas as pd
import numpy as np
df=pd.read_csv("pakistan_used_cars.csv")

print("Missing values:")
print(df.isnull().sum())
initial_rows=len(df)
df=df.dropna(subset=["Brand", "Model"])
df["Registered City"]=df["Registered City"].fillna("unknown")
df["Fuel"]=df["Fuel"].fillna("unknown")
df["Transaction Type"]=df["Transaction Type"].fillna("unknown")
print("Rows Removed:",initial_rows-len(df))

# Formatting
df["Registered City"]=df["Registered City"].str.strip().str.title()
df["Price"]=(df["Price"].astype(str).str.replace("Rs","",regex=False)
             .str.replace(",","",regex=False).str.strip())

# Data types
df["Price"]=pd.to_numeric(df["Price"],errors="coerce").astype("int64")
df["Year"]=pd.to_numeric(df["Year"],errors="coerce").astype("int64")

# Inconsitent Categories
cities_before=df["Registered City"].nunique()
df["Registered City"]=df["Registered City"].replace("Sukkar","Sukkur")
df["Model"]=df["Model"].replace("Corrolla Altis", "Corolla Altis")
cities_after=df["Registered City"].nunique()
print("Different Cities Before:", cities_before)
print("Different Cities After:", cities_after)

# Duplicate
duplicates=df.duplicated().sum()
print("Duplicate Ads:", duplicates)
df=df.drop_duplicates()
print("New Row Count:", len(df))

# Invalid Values
df["KMs Driven"]=pd.to_numeric(df["KMs Driven"], errors="coerce")
invalid_kms=(
    (df["KMs Driven"]<100)|
    (df["KMs Driven"]>1000000)|
    (df["KMs Driven"].isin([12345, 123456]))
)
df.loc[invalid_kms, "KMs Driven"]=np.nan
fake_prices=[12345678, 87654321]
df=df[~df["Price"].isin(fake_prices)]
invalid_ads=(df["Brand"].str.contains("Suzuki|Daihatsu",
                                      case=False, na=False)&(df["Price"]>5000000))

#self check
print("Final Shape:", df.shape)
                                    
                                      
```

- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?

```python
import pandas as pd
import numpy as np
df=pd.read_csv("pakistan_used_cars.csv")

df["Price"]= pd.to_numeric(df["Price"], errors="coerce")
df["Year"]=pd.to_numeric(df["Year"], errors="coerce")
df["Price_lakh"]=df["Price"]/100000
print("(a) Price_lakh")
print(df[["Price", "Price_lakh"]].head())

df["Age"]=2020-df["Year"]
print("(b) Age")
print(df[["Year", "Age"]].head())

print("(c) Top 5 most common Brands:")
print(df["Brand"].value_counts().head(5))

print("(d) Top 5 Cities:")
print(df["Registered City"].value_counts().head(5))

print("(e) Median Price by fuel Type:")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

print("(f) Corolla GLI Median Price_lakh 2010-2018")
corolla=df[df["Model"].str.contains("Corolla GLI",
case=False, na=False)&df["Year"].between(2010, 2018)]
yearly_prices=corolla.groupby("Year")["Price_lakh"].median()
print("Median Price of corolla GLI:")
print(yearly_prices)
if len(yearly_prices)>1:
    slope=np.polyfit(yearly_prices.index.astype(float),
                     yearly_prices.values, 1)[0]
    print("Values Loss  Per Year:", round(-slope, 2), "Lakh Rupees")
    print("Top 5 Most Expensive Ads:")
    print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price"]])
    print("Transaction Type Analysis:")
    transaction_analysis=df.groupby("Transaction Type").agg(Number_of_Ads=("Price", "count"), Median_Price=("Price_lakh", "median"))
```
